```python
numeric_df = df.select_dtypes(include=['number'])
Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR                                                  #for detecting outliers
upper = Q3 + 1.5 * IQR
outliers = ((numeric_df < lower) | (numeric_df > upper))
print(outliers.sum())
```

```python
num_df = df.select_dtypes(include=np.number)
mask = ~((num_df < lower) | (num_df > upper)).any(axis=1)           #for removing outliers
df = df[mask]
```

```python
subjects = ["Math","Reading","Writing"]
marks = [
    df["Math Score"].mean(),
    df["Reading Score"].mean(),
    df["Writing Score"].mean()
]
plt.bar(subjects, marks)
plt.title("Average Subject Scores")                               #for barchart plt.bar()
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.show()
```

```python
plt.plot(df["Math Score"])
plt.title("Math Score")
plt.xlabel("students")
plt.ylabel("Math Marks")                                     #for line chart we use plot()
plt.grid()
plt.show()
```

```python
plt.hist(df["Math Score"],bins=10)
plt.title("Distribution of Math Score")
plt.xlabel("Marks")                                          #for Histogram
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(df["Math Score"],df["Reading Score"])
plt.title("Math vs Reading Score")
plt.xlabel("Math Score")                                            #for Scatterplot we use scatter( )
plt.ylabel("Reading Score")
plt.grid()
plt.show()
```
